## 1. Variables and Python Operators

### 1.1 What is a variable?

A **variable** is a name used to store a value in a program. Python creates a variable when you assign a value to it using `=`.

**Syntax:** `variable_name = value`

```python
# Creating variables
student_name = "Aarav"
age = 21
course_fee = 25000.50

print(student_name)
print(age)
print(course_fee)
```

```python
print('Student Name is:',student_name)
print('Student Age is\n',age)
print('Course Fees is\t',course_fee)
```

```python
usr_inp = input('Enter the value:\n')
print(usr_inp)
print(type(usr_inp))
print(id(usr_inp))
```

### 1.2 Rules for naming variables

- A variable name can contain letters, digits, and underscores (`_`).
- It must begin with a letter or an underscore, not a digit.
- Variable names are case-sensitive (`age` and `Age` are different).
- Python keywords such as `if`, `for`, and `class` cannot be used as variable names.
- Use meaningful names such as `student_marks` instead of unclear names such as `x`.

Valid examples: `name`, `_total`, `student_age`, `marks1`

Invalid examples: `1name`, `student-age`, `class`

```python
# Valid variable names
student_age = 20
total_marks = 450
_name = "Riya"

print(student_age, total_marks, _name)

# Python is case-sensitive
score = 80
Score = 95
print(score)
print(Score)
```

### 1.3 Assigning and updating variables

You can assign values to multiple variables in one statement. You can also update a variable after it has been created.

```python
# Multiple assignment
x, y, z = 10, 20, 30
print(x, y, z)

# Assigning the same value to multiple variables
a = b = c = 100
print(a, b, c)

# Updating a variable
count = 5
count = count + 1
print(count)
```

### 1.4 Arithmetic operators

Arithmetic operators perform mathematical calculations.

| Operator | Meaning | Example | Result |
|---|---|---:|---:|
| `+` | Addition | `10 + 3` | `13` |
| `-` | Subtraction | `10 - 3` | `7` |
| `*` | Multiplication | `10 * 3` | `30` |
| `/` | Division | `10 / 3` | `3.333...` |
| `//` | Floor division | `10 // 3` | `3` |
| `%` | Modulus (remainder) | `10 % 3` | `1` |
| `**` | Exponentiation | `10 ** 3` | `1000` |

```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
```

### 1.5 Comparison operators

Comparison operators compare two values and return either `True` or `False`.

```python
p = 25
q = 20

print(p == q)  # Equal to
print(p != q)  # Not equal to
print(p > q)   # Greater than
print(p < q)   # Less than
print(p >= q)  # Greater than or equal to
print(p <= q)  # Less than or equal to
```

### 1.6 Logical operators

Logical operators combine or reverse conditions.

- `and`: True only when both conditions are True.
- `or`: True when at least one condition is True.
- `not`: Reverses the Boolean result.

```python
age = 22
has_id = True

print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
```

### 1.7 Assignment operators

Assignment operators assign or update values.

```python
balance = 1000

balance += 500   # balance = balance + 500
print("After deposit:", balance)

balance -= 200   # balance = balance - 200
print("After withdrawal:", balance)

balance *= 2
print("After multiplication:", balance)

balance /= 2
print("After division:", balance)
```

### 1.8 Membership and identity operators

- `in` and `not in` check whether a value exists in a sequence.
- `is` and `is not` check whether two references point to the same object. They are not substitutes for `==`, which compares values.

```python
courses = ["Python", "SQL", "Power BI"]

print("SQL" in courses)
print("Tableau" not in courses)

first_list = [1, 2, 3]
second_list = first_list
third_list = [1, 2, 3]

print(first_list == third_list)  # Same values
print(first_list is second_list) # Same object
print(first_list is third_list)  # Different objects
```

### 1.9 Operator precedence

Python follows an order of precedence when an expression contains multiple operators. Parentheses can make the intended order clear.

A simplified order is:
1. Parentheses `()`
2. Exponentiation `**`
3. Multiplication, division, floor division, modulus
4. Addition and subtraction
5. Comparisons
6. `not`, `and`, `or`

Try predicting the output before running the cell.

```python
print(10 + 5 * 2)
print((10 + 5) * 2)
print(2 ** 3 ** 2)
```

## 2. Data Types, Identifiers, and Literals

### 2.1 What is a data type?

A data type defines the kind of value stored in a variable. Common built-in Python types include:

| Data type | Description | Example |
|---|---|---|
| `int` | Whole numbers | `25`, `-10` |
| `float` | Decimal numbers | `98.5`, `3.14` |
| `str` | Text | `"Python"` |
| `bool` | Logical values | `True`, `False` |
| `list` | Ordered, mutable collection | `[10, 20, 30]` |
| `tuple` | Ordered, immutable collection | `(10, 20, 30)` |
| `set` | Collection of unique values | `{10, 20, 30}` |
| `dict` | Key-value pairs | `{"name": "Aarav"}` |
| `NoneType` | Represents no value | `None` |

```python
student_id = 101             # int
percentage = 87.5            # float
student_name = "Aarav"       # str
is_placed = False            # bool
subjects = ["Python", "SQL"] # list
coordinates = (10, 20)       # tuple
unique_scores = {80, 90, 80} # set
student = {"name": "Aarav", "age": 21} # dict
result = None                # NoneType

print(type(student_id))
print(type(percentage))
print(type(student_name))
print(type(is_placed))
print(type(subjects))
print(type(coordinates))
print(type(unique_scores))
print(type(student))
print(type(result))
```

### 2.2 Mutable and immutable types

**Mutable** objects can be changed after creation. Lists, dictionaries, and sets are common mutable types.

**Immutable** objects cannot be changed in place. Integers, floats, strings, Booleans, and tuples are common immutable types.

```python
# A list is mutable
marks = [70, 80, 90]
marks[0] = 75
print(marks)

# A string is immutable: create a new string instead
course = "Python"
new_course = "J" + course[1:]
print(course)
print(new_course)
```

### 2.3 What are identifiers?

An **identifier** is the name given to a variable, function, class, or other program element.

Examples:
- `student_name` — variable identifier
- `calculate_total` — possible function identifier
- `Student` — possible class identifier

Identifier naming rules are the same as the variable naming rules discussed earlier. Python convention commonly uses `snake_case` for variables and functions.

### 2.4 What are literals?

A **literal** is a fixed value written directly in the code.

Examples:
- Integer literal: `100`
- Float literal: `98.6`
- String literal: `"Hello"`
- Boolean literal: `True`
- Special literal: `None`
- Collection literals: `[1, 2]`, `(1, 2)`, `{"a": 1}`

```python
# Examples of literals
number = 100
price = 99.50
message = "Welcome to Python"
active = True
empty_value = None
numbers = [1, 2, 3]

print(number, price, message, active, empty_value, numbers)
```

## 3. Type Casting

### 3.1 What is type casting?

**Type casting** means converting a value from one data type to another.

Common conversion functions:

| Function | Converts to | Example |
|---|---|---|
| `int()` | Integer | `int("25")` |
| `float()` | Float | `float("25.5")` |
| `str()` | String | `str(100)` |
| `bool()` | Boolean | `bool(1)` |
| `list()` | List | `list("abc")` |
| `tuple()` | Tuple | `tuple([1, 2])` |
| `set()` | Set | `set([1, 1, 2])` |

### 3.2 Explicit type casting

Explicit casting happens when the programmer converts a value using a conversion function.

```python
# String to integer
age_text = "21"
age_number = int(age_text)
print(age_number, type(age_number))

# Integer to float
quantity = 5
quantity_float = float(quantity)
print(quantity_float, type(quantity_float))

# Number to string
total = 450
total_text = str(total)
print(total_text, type(total_text))

# List to tuple
items = ["pen", "book", "bag"]
items_tuple = tuple(items)
print(items_tuple, type(items_tuple))
```

### 3.3 Implicit type conversion

In some numeric expressions, Python automatically converts a value to a compatible type. For example, an integer is promoted to a float when combined with a float.

```python
integer_value = 10
float_value = 2.5

result = integer_value + float_value

print(result)
print(type(result))
```

### 3.4 Important type-casting examples and errors

- `int("25")` works because the string contains a valid integer.
- `int("25.5")` raises `ValueError`; convert it to `float` first if a decimal is expected.
- `float("25.5")` works.
- `bool(0)` is `False`, while `bool(1)` is `True`.
- Empty strings and empty collections are false-like; non-empty ones are true-like.

```python
print(int("25"))
print(float("25.5"))
print(int(float("25.5")))  # Decimal part is discarded, not rounded

print(bool(0))
print(bool(1))
print(bool(""))
print(bool("Python"))
print(bool([]))
print(bool([1, 2]))
```

### 3.5 Handling invalid conversions

A conversion can fail if the input is not suitable for the target type. Use `try` and `except` to handle such errors safely.

```python
value = "Python"

try:
    converted_value = int(value)
    print(converted_value)
except ValueError:
    print("Conversion failed: this text cannot be converted to an integer.")
```

## 4. Comments

### 4.1 What are comments?

Comments are notes in code that Python ignores during execution. They help explain the purpose or logic of the program.

A single-line comment begins with `#`.

```python
# This is a single-line comment
student_name = "Riya"  # Store the learner's name

# Display the learner's name
print(student_name)
```

### 4.2 Multi-line comments

Python does not have a dedicated multi-line comment symbol. You can write several single-line comments using `#`.

Triple-quoted strings (`'''...'''` or `"""..."""`) are string literals, not technically comments. They are commonly used as **docstrings** when placed at the beginning of a module, function, or class.

```python
# Calculate the average of three marks.
# Add the marks and divide the total by three.
mark1 = 80
mark2 = 90
mark3 = 85

average = (mark1 + mark2 + mark3) / 3
print("Average:", average)
```

### 4.3 Docstring example

A docstring describes a function, class, or module and can be accessed through the `__doc__` attribute.

```python
def calculate_square(number):
    """Return the square of the given number."""
    return number ** 2

print(calculate_square(5))
print(calculate_square.__doc__)
```

## 5. Mini Practical: Student Marks Calculator

This practical combines variables, data types, operators, type casting, and comments.

**Task:** Store a student's name and three subject marks. Calculate the total and average, then display whether the student has scored at least 40 in every subject.

```python
# Student details
student_name = "Aarav"
python_marks = 85
sql_marks = 78
excel_marks = 92

# Calculate total and average
total_marks = python_marks + sql_marks + excel_marks
average_marks = total_marks / 3

# Check whether all subject marks are at least 40
passed_all = (
    python_marks >= 40
    and sql_marks >= 40
    and excel_marks >= 40
)

print("Student:", student_name)
print("Total marks:", total_marks)
print("Average marks:", round(average_marks, 2))
print("Passed all subjects:", passed_all)
```
